```python
# Simulación de Recepcionista con Lógica de IA Simple
print("--- BIENVENIDO AL HOTEL INTELIGENTE ---")

# Variables iniciales
nombre_cliente = input("Hola, soy tu asistente virtual. ¿Cómo te llamas? ")
sesion_activa = True # Bandera para mantener el ciclo activo

while sesion_activa:
    print(f"\nHola {nombre_cliente}, ¿en qué puedo ayudarte hoy?")
    print("1. Información de Check-in")
    print("2. Calcular presupuesto de Reserva")
    print("3. Reportar una Urgencia")
    print("4. Terminar conversación")

    opcion = input("Selecciona un número (1-4): ")

    if opcion == "1":
        # Lógica de validación con if-else y operadores relacionales
        edad = int(input("Para el registro, ¿qué edad tienes? "))
        if edad >= 18:
            print("IA: Eres mayor de edad. Puedes realizar tu registro ahora mismo.")
        else:
            # Uso de operadores lógicos para condiciones compuestas
            acompanado = input("IA: ¿Vienes con un adulto responsable? (si/no): ").lower()
            if acompanado == "si":
                print("IA: Registro aceptado bajo supervisión de un tutor.")
            else:
                print("IA: Lo siento, por seguridad no puedo registrar a menores solos.")

    elif opcion == "2":
        # Lógica con operadores aritméticos y relacionales [4, 5]
        noches = int(input("¿Cuántas noches piensas quedarte? "))
        precio_noche = 120.50
        total_bruto = noches * precio_noche # Operador de multiplicación

        # IA decide un descuento basado en volumen (operador lógico or)
        if noches > 5 or total_bruto > 500:
            descuento = total_bruto * 0.15 # 15% de descuento
            total_final = total_bruto - descuento
            print(f"IA: ¡Tienes un descuento especial! El total sería ${total_final:.2f}")
        else:
            print(f"IA: El costo total de tu estancia es ${total_bruto:.2f}")

    elif opcion == "3":
        # Clasificación jerárquica con if-elif [5]
        nivel = int(input("Del 1 al 10, ¿qué tan grave es el problema? "))
        if nivel >= 8:
            print("IA: ¡ALERTA! Enviando personal de seguridad de inmediato.")
        elif 4 <= nivel < 8:
            print("IA: Entendido. Un técnico pasará por tu habitación en breve.")
        else:
            print("IA: Tomo nota del inconveniente para resolverlo pronto.")

    elif opcion == "4":
        # Condición de salida para romper el ciclo while [6]
        print(f"IA: Ha sido un gusto atenderte, {nombre_cliente}. ¡Adiós!")
        sesion_activa = False

    else:
        print("IA: No entiendo esa opción. Por favor, elige un número del menú.")
```
